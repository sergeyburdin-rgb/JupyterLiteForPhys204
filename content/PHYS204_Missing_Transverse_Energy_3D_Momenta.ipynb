# PHYS204 — Missing transverse energy from three clusters

**Aim:** Calculate the missing transverse momentum (reported as missing transverse energy, $E_T^{\rm miss}$) from three measured calorimeter clusters. Change the input values and rerun the cells.

**Assumption:** Each cluster is treated as a massless particle (e.g. a photon), so $p_T = E_T = E/\cosh\eta$. All energies are in GeV and azimuthal angles $\phi$ are in radians.

## Formulae

Assume each cluster is a photon (or a massless particle). In natural units ($c=1$), the magnitude of its three-momentum is $p_i=E_i$.

The polar angle $\theta$ is related to pseudorapidity $\eta$ by

$$\theta_i=2\arctan(e^{-\eta_i}).$$

Its unit direction vector is

$$\hat{\mathbf p}_i=(\sin\theta_i\cos\phi_i,\;\sin\theta_i\sin\phi_i,\;\cos\theta_i).$$

The three-momentum is $\mathbf p_i=E_i\hat{\mathbf p}_i$, and $p_{T,i}=\sqrt{p_{x,i}^2+p_{y,i}^2}=E_i/\cosh\eta_i$.

The missing transverse momentum is $\mathbf p_T^{\rm miss}=-\sum_i(p_{x,i},p_{y,i})$, and its magnitude is reported as $E_T^{\rm miss}$. Longitudinal momentum is **not** required to sum to zero at a hadron collider.

## 1. Enter cluster energies and angles

Edit the three rows below and run the notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

# Each row: [energy (GeV), pseudorapidity eta, azimuth phi (radians)]
clusters = np.array([
    [80.0,  0.4,  0.2],
    [65.0, -0.8,  2.5],
    [45.0,  1.1, -1.8],
])

E, eta, phi = clusters.T
```

## 2. Calculate polar angles and unit direction vectors

For each cluster, calculate the polar angle $\theta_i$ in radians and degrees, then the direction vector $(u_x,u_y,u_z)$. The norm of each unit vector should equal 1.

```python
# Polar angles from pseudorapidity
theta = 2 * np.arctan(np.exp(-eta))
theta_deg = np.degrees(theta)

# Unit direction vectors in Cartesian coordinates
ux = np.sin(theta) * np.cos(phi)
uy = np.sin(theta) * np.sin(phi)
uz = np.cos(theta)
unit_vectors = np.column_stack((ux, uy, uz))

print("Cluster   theta (rad)   theta (deg)       ux        uy        uz    |u|")
for i in range(len(E)):
    print(f"{i+1:^7} {theta[i]:12.4f} {theta_deg[i]:13.2f} "
          f"{ux[i]:9.5f} {uy[i]:9.5f} {uz[i]:9.5f} "
          f"{np.linalg.norm(unit_vectors[i]):6.3f}")
```

## 3. Calculate the 3D momentum components

For massless clusters, $p_x=Eu_x$, $p_y=Eu_y$ and $p_z=Eu_z$. Momentum components are shown in GeV/$c$, where numerical values use $c=1$.

```python
# Each cluster three-momentum (GeV/c)
px = E * ux
py = E * uy
pz = E * uz
pT = np.hypot(px, py)

print("Cluster     px (GeV/c)    py (GeV/c)    pz (GeV/c)     pT (GeV/c)")
for i in range(len(E)):
    print(f"{i+1:^7} {px[i]:14.2f} {py[i]:14.2f} "
          f"{pz[i]:14.2f} {pT[i]:14.2f}")
print(f"Total:    {np.sum(px):14.2f} {np.sum(py):14.2f} {np.sum(pz):14.2f}")
```

## 4. Calculate missing transverse momentum and energy

Only the $x$ and $y$ components are used to calculate $E_T^{\rm miss}$; the $z$ component does not enter.

```python
px_miss = -np.sum(px)
py_miss = -np.sum(py)
MET = np.hypot(px_miss, py_miss)
phi_miss = np.arctan2(py_miss, px_miss)

print(f"Missing px = {px_miss:.2f} GeV/c")
print(f"Missing py = {py_miss:.2f} GeV/c")
print(f"Missing transverse energy = {MET:.2f} GeV")
print(f"Missing transverse azimuth = {phi_miss:.3f} rad")
```

## 3. Plot cluster transverse vectors and the missing vector

The missing vector points opposite to the total visible transverse momentum.

```python
fig, ax = plt.subplots(figsize=(7, 7))

for i in range(len(E)):
    ax.arrow(0, 0, px[i], py[i], length_includes_head=True,
             head_width=2.0, head_length=3.0, linewidth=1.5,
             label=f"Cluster {i+1}")

ax.arrow(0, 0, px_miss, py_miss, length_includes_head=True,
         head_width=2.0, head_length=3.0, linewidth=2.5,
         color="black", label=f"Missing $p_T$ = {MET:.1f} GeV")

scale = max(10, np.max(np.abs(np.r_[px, py, px_miss, py_miss])) * 1.3)
ax.set_xlim(-scale, scale)
ax.set_ylim(-scale, scale)
ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(0, color="gray", linewidth=0.8)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel(r"$p_x$ (GeV)")
ax.set_ylabel(r"$p_y$ (GeV)")
ax.set_title("Transverse momenta of three clusters")
ax.grid(True, alpha=0.3)
ax.legend(loc="best")
plt.show()
```
