### 2. Load the Monte Carlo and experimental data
The files are expected to have six columns, ordered as `eta1, phi1, pt1, eta2, phi2, pt2`. Transverse momenta are assumed to be in MeV/$c$.

```python
from pathlib import Path

for filename in ("gamgam_mc.txt", "gamgam_data.txt"):
    if not Path(filename).is_file():
        raise FileNotFoundError(f"Upload {filename} to the notebook folder in JupyterLite first.")

mc = np.genfromtxt("gamgam_mc.txt", unpack=True)  # Simulation data
data = np.genfromtxt("gamgam_data.txt", unpack=True)  # Experimental data

if mc.ndim != 2 or data.ndim != 2 or mc.shape[0] != 6 or data.shape[0] != 6:
    raise ValueError(f"Expected six rows after unpack=True; got MC {mc.shape}, data {data.shape}")

# Pseudorapidity (eta), azimuthal angle (phi), transverse momentum (pt)
mceta1, mcphi1, mcpt1, mceta2, mcphi2, mcpt2 = mc
dataeta1, dataphi1, datapt1, dataeta2, dataphi2, datapt2 = data
print(f"Loaded {mc.shape[1]} MC events and {data.shape[1]} data events")
```

### 3. Plot the first-photon transverse momentum

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.hist(mcpt1 / 1000, bins=30, range=(0, 200), alpha=0.6, label="Monte Carlo")
ax.hist(datapt1 / 1000, bins=30, range=(0, 200), alpha=0.6, label="Experimental data")
ax.set(title="Transverse momentum of the first photon", xlabel=r"$p_{T1}$ (GeV/$c$)", ylabel="Number of events")
ax.legend()
plt.show()
```
